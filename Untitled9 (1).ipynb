```python
# ============================================================
# MINI PROJECT 3: IRIS FLOWER CLUSTERING
# K-Means Clustering (k = 3)
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

from google.colab import files
import pandas as pd
import io
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score


# ------------------------------------------------------------
# 2. UPLOAD IRIS CSV FILE
# ------------------------------------------------------------

print("Please upload your Iris CSV file.")

uploaded = files.upload()

# Automatically detect uploaded filename
filename = list(uploaded.keys())[0]

# Read CSV file
df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("\nDataset uploaded successfully!")
print("File name:", filename)


# ------------------------------------------------------------
# 3. DISPLAY FIRST 5 ROWS
# ------------------------------------------------------------

print("\n========== FIRST 5 ROWS ==========")

print(df.head())


# ------------------------------------------------------------
# 4. DISPLAY DATASET INFORMATION
# ------------------------------------------------------------

print("\n========== DATASET SHAPE ==========")

print(df.shape)


print("\n========== COLUMN NAMES ==========")

print(df.columns.tolist())


print("\n========== DATASET INFORMATION ==========")

df.info()


# ------------------------------------------------------------
# 5. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\n========== MISSING VALUES ==========")

print(df.isnull().sum())


# ------------------------------------------------------------
# 6. SELECT FEATURES FOR CLUSTERING
# ------------------------------------------------------------

# We use only flower measurements.
# Id is NOT used because it has no biological meaning.

features = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm'
]

X = df[features]

print("\n========== FEATURES USED ==========")

print(X.head())


# ------------------------------------------------------------
# 7. APPLY K-MEANS CLUSTERING
# ------------------------------------------------------------

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train K-Means
kmeans.fit(X)

print("\nK-Means model trained successfully!")


# ------------------------------------------------------------
# 8. GET PREDICTED CLUSTERS
# ------------------------------------------------------------

df['Cluster'] = kmeans.labels_

print("\n========== PREDICTED CLUSTERS ==========")

print(df[['Species', 'Cluster']].head(10))


# ------------------------------------------------------------
# 9. DISPLAY CLUSTER CENTERS
# ------------------------------------------------------------

print("\n========== CLUSTER CENTERS ==========")

cluster_centers = pd.DataFrame(
    kmeans.cluster_centers_,
    columns=features
)

print(cluster_centers)


# ------------------------------------------------------------
# 10. VISUALIZE THE CLUSTERS
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    df['PetalLengthCm'],
    df['PetalWidthCm'],
    c=df['Cluster'],
    cmap='viridis',
    s=50
)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Iris Flower Clusters using K-Means")

plt.show()


# ------------------------------------------------------------
# 11. COMPARE TRUE SPECIES WITH PREDICTED CLUSTERS
# ------------------------------------------------------------

print("\n========== TRUE SPECIES VS PREDICTED CLUSTERS ==========")

comparison = pd.crosstab(
    df['Species'],
    df['Cluster']
)

print(comparison)


# ------------------------------------------------------------
# 12. CALCULATE ADJUSTED RAND INDEX
# ------------------------------------------------------------

# Convert species names into numerical labels
true_labels = pd.factorize(df['Species'])[0]

# Calculate ARI
ari = adjusted_rand_score(
    true_labels,
    df['Cluster']
)

print("\n========== ADJUSTED RAND INDEX ==========")

print("ARI Score:", ari)


# ------------------------------------------------------------
# 13. FINAL DATASET
# ------------------------------------------------------------

print("\n========== FINAL DATASET ==========")

print(
    df[
        [
            'SepalLengthCm',
            'SepalWidthCm',
            'PetalLengthCm',
            'PetalWidthCm',
            'Species',
            'Cluster'
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 14. FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n========================================")
print("       IRIS CLUSTERING PROJECT")
print("========================================")

print("Dataset:", filename)
print("Number of data samples:", len(df))
print("Number of clusters:", kmeans.n_clusters)
print("Features used:", len(features))
print("ARI Score:", ari)

print("\nProject completed successfully!")
```
